```python
import matplotlib.pyplot as plt
import networkx as nx

G = nx.DiGraph()

relationships = [
    ("Priya", "Student", "is a"),
    ("Student", "AI", "studies"),
    ("Student", "Data Mining", "studies"),
    ("Student", "CSE", "belongs to"),
    ("AI", "Subject", "is a"),
    ("Data Mining", "Subject", "is a"),
    ("CSE", "Department", "is a"),
]

for source, target, relation in relationships:
    G.add_edge(source, target, relation=relation)

pos = {
    "Priya": (0, 3),
    "Student": (0, 1.8),
    "AI": (-1.8, 0.4),
    "Data Mining": (0, 0.4),
    "CSE": (1.8, 0.4),
    "Subject": (-0.9, -1),
    "Department": (1.8, -1),
}

plt.figure(figsize=(10, 8))

nx.draw_networkx_nodes(
    G,
    pos,
    node_color="lightblue",
    edgecolors="black",
    node_size=2800,
    linewidths=1.5,
)

nx.draw_networkx_labels(
    G, pos, font_size=9, font_weight="bold", font_family="sans-serif"
)

nx.draw_networkx_edges(
    G,
    pos,
    edge_color="black",
    arrowstyle="-|>",
    arrowsize=15,
    node_size=2800,
    width=1.2,
)

edge_labels = nx.get_edge_attributes(G, "relation")
nx.draw_networkx_edge_labels(
    G,
    pos,
    edge_labels=edge_labels,
    font_size=8,
    font_color="black",
    rotate=False,
)

plt.axis("off")
plt.tight_layout()
plt.show()
```
